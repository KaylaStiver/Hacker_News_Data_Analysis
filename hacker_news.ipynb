# Hacker News Analysis

By analyzing data from user-generated posts, this study identifies trends in popular Hacker News topics to determine whether "Ask Hacker News" or "Show Hacker News" posts are more popular amongst users. In addition, peak user activity time frames are identified from raw data.

## Data at a Glance
This dataset was scraped from the Hacker News site over the course of a year, ending in September 2016.
Sourced from Kaggle, for more information please refer to the link below:
* [Hacker News Posts](https://www.kaggle.com/datasets/hacker-news/hacker-news-posts)

Now, to print the first few rows of the data.

```python
# reading csv and printing first 5 rows
import csv 
with open('hacker_news.csv', mode='r', newline='', encoding='utf-8') as file:
    hn = csv.reader(file)
    header = next(hn) # saving header with column names row separately
    hn_list = list(hn)

print(header)
for index, row in enumerate(hn_list):
    if index >= 5:
        break
    print("\n")
    print(row)
    
```

## Analysis

Columns detail basic information about the post such as the title, URL, who created the post, when it was created, and the number of comments.

Now to sort the data into a more useful format. The data will be sorted into "Ask HN (Hacker News)", "Show HN (Hacker News)", and other posts. The other posts are irrelevant for our purposes and need to be removed. In addition, the number of comments will be tallied.

```python
ask_posts = []
show_posts = []
other_posts = []

for row in hn_list:
    title = row[1] # index of title column
    if title.lower().startswith("ask hn"):
        ask_posts.append(row)
    elif title.lower().startswith("show hn"):
        show_posts.append(row)
    else:
        other_posts.append(row)

print(f"Ask posts have: {len(ask_posts)} comments.")
print(f"Show posts have: {len(show_posts)} comments.")
print(f"Other posts have: {len(other_posts)} comments.")
```

We are only interested in the "Ask HN" and "Show HN" posts. The number of comments per post type is calculated below for each relevant category to give a clearer metric of which post type generates more user activity.

```python
total_ask_comments = 0
total_show_comments = 0

for row in ask_posts:
    comments = row[4] # index of num_comments column
    total_ask_comments += int(comments)

avg_ask_comments = total_ask_comments/len(ask_posts) # total comments divided by number of posts
print(f"Ask posts average: {avg_ask_comments}")

for row in show_posts: 
    comments = row[4]
    total_show_comments += int(comments) # calculating total comments for the entire "Ask HN" category

avg_show_comments = total_show_comments/len(show_posts)
print(f"Show posts average: {avg_show_comments}")
```

On average, "Ask HN" posts generate 14 comments per post. Compared to "Show HN' which generates less user activity, on average, with 10 comments per post.

To optimize user acitivty on a post, it can be concluded it should be of the "Ask HN" category. However, timing of the post would also affect the user activity. Assumedly, a post created during the day for the majority of users would get more acitvity than one created at a low traffic timeframe.

Let's find out when Hacker News tends to be the most active. We can do this by analzying the `created_at` column. A frequency dictionary will be created for the "Ask HN" posts tallying the hours by which each post was created and then finding the average per hour.

```python
import datetime as dt

result_list = []
for row in ask_posts:
    temp_list = []
    created_on = row[6] # index of created_at column
    num_comments = row[4] # index of num_comments column
    result_list.append([row[6], int(row[4])])

counts_by_hour = {} # number of posts by hour
comments_by_hour = {} # total comments per hour
for row in result_list:
    date = dt.datetime.strptime(row[0], "%m/%d/%Y %H:%M") # formatting string into datetime object
    hour = dt.datetime.strftime(date, "%H") # taking only the hour from the datetime 
    if hour not in counts_by_hour:
        counts_by_hour[hour] = 1
        comments_by_hour[hour] = row[1]
    else:
        counts_by_hour[hour] += 1
        comments_by_hour[hour] += row[1]
```

```python
avg_by_hour = []

for hour in counts_by_hour:
    total_comments = int(comments_by_hour[hour]) # total comments per hour
    num_posts = counts_by_hour[hour] # number of posts by hour
    avg_comments = total_comments/num_posts
    avg_by_hour.append([hour, avg_comments])
print(avg_by_hour)
```

The dictionary above has the average number of comments per hour, with the hour being the key and the average number of comments as the value. However, this is quite hard to read, to remedy this the key and its values will be reversed so the entries will be sorted by the highest number of average comments per hour.

```python
swap_avg_by_hour = []
for row in avg_by_hour:
    swap_avg_by_hour.append([row[1],row[0]]) # swaps the key and its values within the dictionary
print(swap_avg_by_hour)
```

The dictionary has been swapped above. Now, the top 5 most active hours by average number of comments can be found.

```python
sorted_swap = sorted(swap_avg_by_hour, reverse=True)
format_date = "%H"
print("Top 5 Hours for Ask Posts Comments")
for index, row in enumerate(sorted_swap):
    if index >= 5:
        break
    date = dt.datetime.strptime(row[1], format_date)
    dt_hour = date.strftime("%H")
    print(f"{dt_hour}:00: {row[0]:.2f} average comments per post.")
```

The hours listed above are in Eastern Standard Time (EST). The most active hour on Hacker News for "Ask HN" posts is 15:00 EST per average comments per post.

## Conclusion

The most engaging type of post on Hacker News belongs to the "Ask HN" topic with 14 average comments per post. Among the "Ask HN" posts the most popular time to generate the most user activity is 15:00 EST.
